**Question 2:** In how many different states do the customers live in?

```python
import csv

# open file and init vars
# evaluate the contents of the file
with open(datafile,'r') as fl:
  csvreader = csv.reader(fl,delimiter=',')
  #States = [state for state in csvreader if state]
  States = []
  for state in csvreader:
    if state[4].strip().upper() not in States:
      States.append(state[4].upper())
  print(f"Number of states { len(States)-1 }")
```

**Question 3** What is the state with most customers?

```python
# open file and init vars
# evaluate the contents of the file
with open(datafile,'r') as fl:
  csvreader = csv.reader(fl,delimiter=',')
  States = [state[4] for state in csvreader if state[4] != 'State']
  print(f"State with most customers {max(set(States), key = States.count)}")
  #
```

**Question 4** What is the state with the least customers?

```python
# open file and init vars
# evaluate the contents of the file
import pandas as pd
with open(datafile,'r') as fl:
  csvreader = csv.reader(fl,delimiter=',')
  States = [state[4] for state in csvreader if state[4] != 'State']
  data = pd.DataFrame(States)
  data.columns = ['State']
  contStat = data['State'].value_counts(ascending=True)
  minimo = contStat.iloc[0]
  minStat = contStat[contStat == minimo]
  print(f"States with least customers {minStat}")
```

**Question 5:** What is the most common last name?

```python
# open file and init vars
# evaluate the contents of the file
with open(datafile,'r') as fl:
  csvreader = csv.reader(fl,delimiter=',')
  LastName = [state[2] for state in csvreader if state[2] != 'Last Name']
  print(f"Most common last name { max(set(LastName), key = LastName.count) }")
```
